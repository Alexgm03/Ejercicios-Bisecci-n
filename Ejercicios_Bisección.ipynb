# Ejercicio 1

Resolver usando el método de bisección:

$$
f(x)=x^3-7x^2+14x-6
$$

```python
def f(x):
    return x**3 - 7*x**2 + 14*x - 6

def biseccion(a, b, tol):

    while (b - a)/2 > tol:

        c = (a + b)/2

        if f(c) == 0:
            return c

        elif f(a)*f(c) < 0:
            b = c

        else:
            a = c

    return (a + b)/2
```

## a) Intervalo [0,1]

```python
raiz = biseccion(0,1,1e-2)

print(raiz)
```

## b) Intervalo [1,3.2]

```python
raiz = biseccion(1,3.2,1e-2)

print(raiz)
```

## c) Intervalo [3.2,4]

```python
raiz = biseccion(3.2,4,1e-2)

print(raiz)
```

# Ejercicio Aplicado 1

$$
V = L\left[0.5\pi r^2 - r^2\arcsin\left(\frac{h}{r}\right) - h\sqrt{r^2-h^2}\right]
$$

```python
import math

L = 10
r = 1
V = 12.4

def volumen(h):

    return L * (
        0.5 * math.pi * r**2
        - r**2 * math.asin(h/r)
        - h * math.sqrt(r**2 - h**2)
    ) - V

a = 0
b = 1
tol = 1e-2

while (b - a)/2 > tol:

    c = (a + b)/2

    if volumen(c) == 0:
        break

    elif volumen(a) * volumen(c) < 0:
        b = c

    else:
        a = c

print("Profundidad:", (a+b)/2)
```

# Ejercicio Aplicado 2

$$
s(t)=s_0-\frac{mg}{k}t+\frac{m^2g}{k^2}\left(1-e^{-kt/m}\right)
$$

```python
import numpy as np

g = 9.81
s0 = 300
m = 0.25
k = 0.1

def altura(t):

    return s0 - ((m*g)/k)*t + ((m**2*g)/(k**2))*(1 - np.exp((-k*t)/m))

a = 0
b = 20
tol = 1e-1

while (b - a)/2 > tol:

    c = (a + b)/2

    if altura(c) == 0:
        break

    elif altura(a) * altura(c) < 0:
        b = c

    else:
        a = c

print("Tiempo:", (a+b)/2)
```
